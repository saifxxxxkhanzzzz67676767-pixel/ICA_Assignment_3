```python
import pandas as pd
```

```python
file_path = "teaching_prices_messy (1).csv"
```

```python
raw = pd.read_csv(r"C:\Users\MCT\Downloads\teaching_prices_messy (1).csv")
```

```python
df = raw.copy()
```

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")
```

```python
df["item"] = df["item"].astype("string").str.strip().str.title()
```

```python
df["item"] = df["item"].astype("string").str.strip().str.title()
```

```python
df["price"] = pd.to_numeric(
    df["price"].astype("string").str.replace(",", "", regex=False),
    errors="coerce"
)
```

```python
duplicate_rows = df[df.duplicated(keep="first")].copy()
```

```python
df = df.drop_duplicates().copy()
```

```python
flagged = (
    ((df["item"] == "Staple A") & (df["date"] == pd.Timestamp("2026-01-15"))
     & (df["city"] == "City 2") & (df["price"] == 1250))
    | ((df["item"] == "Sweetener D") & (df["date"] == pd.Timestamp("2026-01-22"))
       & (df["city"] == "City 1") & (df["price"] == 9999))
    | ((df["item"] == "Oil B") & (df["date"] == pd.Timestamp("2026-02-05"))
       & (df["city"] == "City 1") & (df["price"] == 108.5))
)
```

```python
df["flag_reason"] = ""
df.loc[flagged, "flag_reason"] = "Unusual observation; verify against source"
```

```python
clean = df.loc[~flagged].copy()
```

```python
print("RAW ROWS:", len(raw))
```

```python
print("ROWS AFTER EXACT-DEDUPE:", len(df))
print("EXACT DUPLICATES REMOVED:", len(duplicate_rows))
print("MISSING PRICE VALUES:", int(df["price"].isna().sum()))
print("FLAGGED OBSERVATIONS:")
print(df.loc[flagged, ["date", "item", "city", "price", "flag_reason"]].to_string(index=False))
```

```python
summary = (
    clean.groupby("item")["price"]
    .agg(count="count", mean="mean", median="median")
    .round(2)
```

```python
print("\nMEAN AND MEDIAN BY ITEM (PRIMARY CLEANED DATA):")
print(summary)
```
